```python
import pandas as pd
import spacy
import os
```

```python
nlp = spacy.load("pl_core_news_sm")
```

```python
folder = " "
os.chdir(folder)
```

```python
metrics = pd.read_csv("metrics_final.csv", index_col=0, encoding="utf-8")
```

```python
def calculate_TTR(doc):

    types = set([token.text.lower() for token in doc])
    result = len(types)/len(doc)

    return result
```

```python
def calculate_lemmatized_TTR(doc):
    types = set([token.lemma for token in doc])
    result = len(types)/len(doc)

    return result
```

```python
def MATTR(text, step):
    doc = nlp(text)
    doc = [token for token in doc if not token.is_punct and not token.is_space]
    ttrs = []
    last_index = len(doc) - step + 1

    if len(doc) <= step:
        return(calculate_TTR(doc))

    for i in range(0, last_index):
        window = doc[i:i+step]
        ttrs.append(calculate_TTR(window))

    return sum(ttrs)/len(ttrs)
```

```python
def MATTR_Lemmatized(text, step):
    doc = nlp(text)
    doc = [token for token in doc if not token.is_punct and not token.is_space]
    ttrs = []
    last_index = len(doc) - step + 1

    if len(doc) <= step:
        return(calculate_lemmatized_TTR(doc))

    for i in range(0, last_index):
        window = doc[i:i+step]
        ttrs.append(calculate_lemmatized_TTR(window))

    return sum(ttrs)/len(ttrs)
```

```python
metrics["MATTR"] = metrics.apply(lambda x: MATTR(x["text"], step=100), axis=1)
```

```python
metrics["Lemmatized_MATTR"] = metrics.apply(lambda x: MATTR_Lemmatized(x["text"], step=100), axis=1)
```

```python
metrics.to_csv("metrics_with_mattr.csv", encoding="utf-8")
```
